```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import roc_curve, auc
import numpy as np


# helper
def compute_means(df):
    df["scores"] = pd.to_numeric(df["scores"], errors="coerce")
    return df.groupby("source")["scores"].mean().dropna()


# compute per source means
cancer_means  = compute_means(cancer_df)
t1d_means     = compute_means(t1d_df)
sle_means     = compute_means(sle_df)
control_means = compute_means(merged_control_df)


# ============================================================
# 1) ONE FIGURE: boxplots for all 4 groups
# ============================================================

plt.figure(figsize=(10,6))
plt.boxplot(
    [
        cancer_means.values,
        control_means.values,
        t1d_means.values,
        sle_means.values,
    ],
    labels=["Cancer", "Control", "T1D", "SLE"]
)
plt.ylabel("Mean score")
plt.title("Mean score per source for Cancer, Control, T1D, SLE")
plt.grid(True, linestyle=":")
plt.tight_layout()
plt.show()


# ============================================================
# 2) ONE FIGURE: ROC with 3 comparisons
# ============================================================

plt.figure(figsize=(7,7))

def add_roc(pos, neg, name):
    y_true = np.concatenate([np.ones(len(pos)), np.zeros(len(neg))])
    y_scores = np.concatenate([pos.values, neg.values])
    fpr, tpr, _ = roc_curve(y_true, y_scores)
    roc_auc = auc(fpr, tpr)
    plt.plot(fpr, tpr, linewidth=2, label=f"{name} (AUC={roc_auc:.3f})")


add_roc(cancer_means, control_means, "Cancer vs Control")
add_roc(cancer_means, t1d_means, "Cancer vs T1D")
add_roc(cancer_means, sle_means, "Cancer vs SLE")

plt.plot([0,1], [0,1], linestyle=":", color="black")
plt.xlabel("False positive rate")
plt.ylabel("True positive rate")
plt.title("ROC curves using per source mean scores")
plt.legend()
plt.grid(True, linestyle=":")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Assume your three dataframes already exist:
# summary_long, summary_sle, summary_t1d

# Add group labels to the two without group info
summary_sle["group"] = "SLE"
summary_t1d["group"] = "T1D"

# Concatenate everything
summary_all = pd.concat(
    [summary_long, summary_sle, summary_t1d],
    ignore_index=True
)

# Plot boxplot
plt.figure(figsize=(8, 6))
summary_all.boxplot(column="inv_logit_mean", by="group")
plt.title("inv_logit_mean by group")
plt.suptitle("")  # Remove automatic title
plt.xlabel("Group")
plt.ylabel("inv_logit_mean")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
from sklearn.metrics import roc_curve, auc
import matplotlib.pyplot as plt

# Assume summary_all has columns: group, inv_logit_mean
# and groups: Cancer, Control, SLE, t1d

def make_roc(df, pos_group, neg_group):
    df_pair = df[df["group"].isin([pos_group, neg_group])].copy()
    df_pair["label"] = (df_pair["group"] == pos_group).astype(int)
    y_true = df_pair["label"]
    y_score = df_pair["inv_logit_mean"]
    fpr, tpr, _ = roc_curve(y_true, y_score)
    return fpr, tpr, auc(fpr, tpr)

# Cancer vs Control
fpr_cc, tpr_cc, auc_cc = make_roc(summary_all, "Cancer", "Control")

# Cancer vs SLE
fpr_csle, tpr_csle, auc_csle = make_roc(summary_all, "Cancer", "SLE")

# Cancer vs t1d
fpr_ct1d, tpr_ct1d, auc_ct1d = make_roc(summary_all, "Cancer", "T1D")

# Plot
plt.figure(figsize=(7, 6))
plt.plot(fpr_cc, tpr_cc, label=f"Cancer vs Control AUC={auc_cc:.3f}")
plt.plot(fpr_csle, tpr_csle, label=f"Cancer vs SLE AUC={auc_csle:.3f}")
plt.plot(fpr_ct1d, tpr_ct1d, label=f"Cancer vs t1d AUC={auc_ct1d:.3f}")

plt.plot([0, 1], [0, 1], linestyle="--", color="gray")

plt.xlabel("False positive rate")
plt.ylabel("True positive rate")
plt.title("Three ROC curves")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import precision_recall_curve, average_precision_score

# df must contain: df["group"], df["inv_logit_mean"]

# Convert group labels to binary target
summary_long["label"] = (summary_long["group"] == "Cancer").astype(int)

y_true = summary_long["label"].values
y_score = summary_long["inv_logit_mean"].values

# Compute Precision-Recall curve
precision, recall, thresholds = precision_recall_curve(y_true, y_score)
auprc = average_precision_score(y_true, y_score)

print("AUPRC =", auprc)

# Plot
plt.figure(figsize=(6,5))
plt.plot(recall, precision, linewidth=2)
plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title(f"Precision-Recall Curve (AUPRC = {auprc:.3f})")
plt.grid(True)
plt.show()
```
