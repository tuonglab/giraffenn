```python
from scipy.stats import mannwhitneyu
import statsmodels.stats.multitest as smm
import numpy as np

groups_to_compare = ["Control", "SLE", "T1D", "COVID"]
pvals = []

for g in groups_to_compare:
    a = summary_all.loc[summary_all.group == "Cancer", "inv_logit_mean"]
    b = summary_all.loc[summary_all.group == g, "inv_logit_mean"]
    stat, p = mannwhitneyu(a, b, alternative="greater")
    pvals.append(p)

# FDR correction
pvals_corrected = smm.multipletests(pvals, method="fdr_by")[1]

def p_to_stars(p):
    if p < 1e-4:
        return "****"
    elif p < 1e-3:
        return "***"
    elif p < 1e-2:
        return "**"
    elif p < 0.05:
        return "*"
    else:
        return "ns"   # not significant

stars = [p_to_stars(p) for p in pvals_corrected]
pvals_corrected
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(14, 8))

# Draw boxplot
sns.boxplot(
    data=summary_all,
    x="group",
    y="inv_logit_mean",
    showcaps=True,
    boxprops={'facecolor':'none', 'edgecolor':'black'},
    showfliers=False,
    whiskerprops={'color':'black'},
    medianprops={'color':'black'}
)

# Draw swarmplot
sns.swarmplot(
    data=summary_all,
    x="group",
    y="inv_logit_mean",
    size=4,
    color="black",
    alpha=0.7
)

# x positions corresponding to each group
x_positions = {
    "Cancer": 0,
    "Control": 1,
    "SLE": 2,
    "T1D": 3,
    "COVID": 4
}

# y-position to draw brackets
max_y = summary_all["inv_logit_mean"].max()
baseline = max_y + 0.005
step = 0.01   # vertical spacing between annotations

# Add star annotations for Cancer vs each group
for i, g in enumerate(groups_to_compare):
    x1 = x_positions["Cancer"]
    x2 = x_positions[g]
    y = baseline + i * step
    star = stars[i]

    # draw bracket
    plt.plot([x1, x1, x2, x2],
             [y - 0.002, y, y, y - 0.002],
             color="black", linewidth=1)

    # add star text
    plt.text((x1 + x2) / 2, y + 0.002, star, ha="center", fontsize=12)

plt.title("inv_logit_mean by group (Boxplot + Swarmplot + Significance Stars)")
plt.xlabel("Group")
plt.ylabel("inv_logit_mean")
plt.tight_layout()

plt.savefig("inv_logit_mean_box_swarm_stars.pdf", format="pdf", dpi=300)
plt.show()
plt.close()
```

```python
import numpy as np

def bootstrap_diff_in_means(df, group_col, value_col,
                            g1, g2, n_boot=10000, ci=0.95,
                            random_state=None):
    rng = np.random.default_rng(random_state)

    x = df.loc[df[group_col] == g1, value_col].values
    y = df.loc[df[group_col] == g2, value_col].values

    obs_diff = x.mean() - y.mean()

    boot_diffs = []
    for _ in range(n_boot):
        xb = rng.choice(x, size=len(x), replace=True)
        yb = rng.choice(y, size=len(y), replace=True)
        boot_diffs.append(xb.mean() - yb.mean())

    alpha = 1 - ci
    lower = np.quantile(boot_diffs, alpha / 2)
    upper = np.quantile(boot_diffs, 1 - alpha / 2)

    return {
        "group1": g1,
        "group2": g2,
        "obs_diff": obs_diff,
        f"ci_{int(ci*100)}_lower": lower,
        f"ci_{int(ci*100)}_upper": upper
    }

# Example: Cancer vs Control, SLE, T1D, COVID
pairs = [("Cancer", "Control"),
         ("Cancer", "SLE"),
         ("Cancer", "T1D"),
         ("Cancer", "COVID")]

results_ci = []
for g1, g2 in pairs:
    res = bootstrap_diff_in_means(summary_all,
                                  group_col="group",
                                  value_col="inv_logit_mean",
                                  g1=g1, g2=g2,
                                  n_boot=100000,
                                  ci=0.95,
                                  random_state=0)
    results_ci.append(res)

import pandas as pd
df_ci = pd.DataFrame(results_ci)
print(df_ci)
```

```python
# Correctly define the comparison labels in the order they will be drawn (bottom-up)
df_ci["comparison"] = [
    "Cancer vs Control",  # i=0 (bottom)
    "Cancer vs SLE",      # i=1
    "Cancer vs T1D",      # i=2
    "Cancer vs COVID"     # i=3 (top)
]

# Size of the plot
plt.figure(figsize=(8, 5))

# Iterate through rows to draw CI lines and point estimates
for i, row in df_ci.iterrows():
    # Draw CI line
    plt.plot([row["ci_95_lower"], row["ci_95_upper"]], [i, i],
             color="black", linewidth=2)
    
    # Draw point estimate
    plt.plot(row["obs_diff"], i, "o", color="black")

# Draw zero line for significance reference
plt.axvline(0, color="red", linestyle="--", linewidth=1)

# Apply Y-ticks
plt.yticks(range(len(df_ci)), df_ci["comparison"], fontsize=12)
plt.xlabel("Difference in means (Cancer minus comparator)", fontsize=12)
plt.title("Bootstrap 95 percent confidence intervals for group differences",
          fontsize=14)

plt.tight_layout()
plt.savefig("bootstrap_forest_plot.pdf", dpi=300)
plt.show()
```

```python
# ----------------------------------------------------------
# 2) ROC curves saved as PDF
# ----------------------------------------------------------

from sklearn.metrics import auc, roc_curve


def make_roc(df, pos_group, neg_group):
    df_pair = df[df["group"].isin([pos_group, neg_group])].copy()
    df_pair["label"] = (df_pair["group"] == pos_group).astype(int)
    y_true = df_pair["label"]
    y_score = df_pair["inv_logit_mean"]
    fpr, tpr, _ = roc_curve(y_true, y_score)
    return fpr, tpr, auc(fpr, tpr)

# Cancer vs Control
fpr_cc, tpr_cc, auc_cc = make_roc(summary_all, "Cancer", "Control")

# Cancer vs SLE
fpr_csle, tpr_csle, auc_csle = make_roc(summary_all, "Cancer", "SLE")

# Cancer vs T1D
fpr_ct1d, tpr_ct1d, auc_ct1d = make_roc(summary_all, "Cancer", "T1D")

# Cancer vs COVID
fpr_covid, tpr_covid, auc_covid = make_roc(summary_all, "Cancer", "COVID")

plt.figure(figsize=(7, 6))

plt.plot(fpr_cc, tpr_cc, label=f"Cancer vs Control AUC={auc_cc:.3f}")
plt.plot(fpr_csle, tpr_csle, label=f"Cancer vs SLE AUC={auc_csle:.3f}")
plt.plot(fpr_ct1d, tpr_ct1d, label=f"Cancer vs T1D AUC={auc_ct1d:.3f}")
plt.plot(fpr_covid, tpr_covid, label=f"Cancer vs COVID AUC={auc_covid:.3f}")

plt.plot([0, 1], [0, 1], linestyle="--", color="gray")

plt.xlabel("False positive rate")
plt.ylabel("True positive rate")
plt.title("ROC curves for Cancer vs other groups")
plt.legend()
plt.tight_layout()

plt.savefig("roc_four_groups.pdf", format="pdf", dpi=300)
plt.show()
plt.close()
```

```python
import numpy as np
import pandas as pd
from scipy.stats import mannwhitneyu
from sklearn.metrics import roc_auc_score, roc_curve, confusion_matrix, accuracy_score, recall_score, precision_score
from statsmodels.stats.multitest import multipletests

df = summary_all.copy()

cancer_scores = df.loc[df["group"] == "Cancer", "inv_logit_mean"]
mean_cancer = cancer_scores.mean()

ratios = []

for g in sorted(df["group"].unique()):
    if g == "Cancer":
        continue
    scores_g = df.loc[df["group"] == g, "inv_logit_mean"]
    mean_g = scores_g.mean()
    ratio = mean_cancer / mean_g if mean_g != 0 else np.nan
    ratios.append({"comparison": f"Cancer_vs_{g}", "mean_cancer": mean_cancer,
                   "mean_other": mean_g, "ratio_cancer_over_other": ratio})

# Cancer vs pooled non-cancer
non_cancer_scores = df.loc[df["group"] != "Cancer", "inv_logit_mean"]
mean_non_cancer = non_cancer_scores.mean()
ratio_pooled = mean_cancer / mean_non_cancer if mean_non_cancer != 0 else np.nan

ratios.append({"comparison": "Cancer_vs_all_non_cancer",
               "mean_cancer": mean_cancer,
               "mean_other": mean_non_cancer,
               "ratio_cancer_over_other": ratio_pooled})

ratios_df = pd.DataFrame(ratios)
print("\nRatios of means (Cancer specific evidence)")
print(ratios_df)
```

```python

# Binary labels
y = (df["group"] == "Cancer").astype(int).values
scores = df["inv_logit_mean"].values

# AUC
auc = roc_auc_score(y, scores)
print(f"\nROC AUC for Cancer vs all others: {auc:.3f}")

# ROC curve to find threshold that maximizes Youden's J (sens + spec - 1)
fpr, tpr, thresholds = roc_curve(y, scores)

youden_j = tpr - fpr
best_idx = np.argmax(youden_j)
best_thresh = thresholds[best_idx]

print(f"Best threshold (Youden J): {best_thresh:.4f}")
print(f"At this threshold: sens {tpr[best_idx]:.3f}, spec {1 - fpr[best_idx]:.3f}")

# Confusion matrix and metrics at best threshold
y_pred = (scores >= best_thresh).astype(int)
tn, fp, fn, tp = confusion_matrix(y, y_pred).ravel()

sensitivity = tp / (tp + fn) if (tp + fn) > 0 else np.nan  # recall for Cancer
specificity = tn / (tn + fp) if (tn + fp) > 0 else np.nan
ppv = tp / (tp + fp) if (tp + fp) > 0 else np.nan          # precision
npv = tn / (tn + fn) if (tn + fn) > 0 else np.nan
accuracy = accuracy_score(y, y_pred)

print("\nMetrics at chosen threshold")
print(f"TP: {tp}, FP: {fp}, TN: {tn}, FN: {fn}")
print(f"Sensitivity (Cancer) : {sensitivity:.3f}")
print(f"Specificity (non-Ca) : {specificity:.3f}")
print(f"PPV (precision)      : {ppv:.3f}")
print(f"NPV                  : {npv:.3f}")
print(f"Accuracy             : {accuracy:.3f}")
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import roc_curve, auc

df2 = summary_all.copy()
df2["label"] = (df2["group"] == "Cancer").astype(int)

y_true = df2["label"].values
y_score = df2["inv_logit_mean"].values

fpr, tpr, thresholds = roc_curve(y_true, y_score)
roc_auc = auc(fpr, tpr)

# Compute Youden J
youden_j = tpr - fpr
idx = youden_j.argmax()

best_thresh = thresholds[idx]
best_sens = tpr[idx]
best_spec = 1 - fpr[idx]

plt.figure(figsize=(7, 6))
plt.plot(fpr, tpr, label=f"ROC AUC = {roc_auc:.3f}")
plt.scatter(fpr[idx], tpr[idx], color="red", s=80,
            label=f"Youden point (thr={best_thresh:.3f})")

plt.plot([0, 1], [0, 1], linestyle="--", color="gray")
plt.xlabel("False positive rate")
plt.ylabel("True positive rate")
plt.title("ROC curve with Youden optimal threshold")
plt.legend()
plt.tight_layout()
plt.savefig("roc_curve_cancer_vs_non_cancer.pdf", format="pdf", dpi=300)
plt.show()


import seaborn as sns
import numpy as np

preds = (y_score >= best_thresh).astype(int)

TP = ((preds == 1) & (y_true == 1)).sum()
TN = ((preds == 0) & (y_true == 0)).sum()
FP = ((preds == 1) & (y_true == 0)).sum()
FN = ((preds == 0) & (y_true == 1)).sum()

cm = np.array([[TN, FP],
               [FN, TP]])

plt.figure(figsize=(10, 6))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
            xticklabels=["Pred non cancer", "Pred cancer"],
            yticklabels=["True non cancer", "True cancer"])

plt.title("Confusion matrix at Youden optimal threshold")
plt.tight_layout()
plt.savefig("confusion_matrix_cancer_vs_non_cancer.pdf", format="pdf", dpi=300)
plt.show()

import matplotlib.pyplot as plt

metrics = {
    "Sensitivity": best_sens,
    "Specificity": best_spec,
    "PPV": TP / (TP + FP),
    "NPV": TN / (TN + FN),
    "Accuracy": (TP + TN) / (TP + FP + TN + FN)
}

plt.figure(figsize=(8, 5))
plt.bar(metrics.keys(), metrics.values(), color="lightgreen", edgecolor="black")
plt.ylim(0, 1)
plt.ylabel("Metric value")
plt.title("Diagnostic performance at Youden optimal threshold")
plt.tight_layout()
plt.savefig("diagnostic_performance_cancer_vs_non_cancer.pdf", format="pdf", dpi=300)
plt.show()

```
